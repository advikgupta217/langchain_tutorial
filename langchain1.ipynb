```python
from langchain_groq import ChatGroq

llm = ChatGroq(
    model="openai/gpt-oss-20b",
    temperature=0.6
)

response = llm.invoke(
    "I want to open a restaurant for Italian food. "
    "Suggest only one fancy name for this."
)

print(response.content)
```

```python
from langchain_core.prompts import PromptTemplate

prompt_template_name = PromptTemplate(
    input_variables = ['cuisine'],
    template = "I want to open a restaurant for {cuisine} food. Suggest a fancy name for this."
)

prompt = prompt_template_name.format(cuisine="Italian")
print(prompt)
```

```python
#Create a chain so that we don't have to tell the prompt again and again, just mention the cuisine type.
from langchain_core.prompts import PromptTemplate

prompt_template_name = PromptTemplate(
    input_variables=["cuisine"],
    template="I want to open a restaurant for {cuisine} food. Suggest only one fancy name for this."
)

chain = prompt_template_name | llm

response = chain.invoke({"cuisine": "French"})

print(response.content)
```

```python
from langchain_core.prompts import PromptTemplate

prompt_template_name = PromptTemplate(
    input_variables=["cuisine"],
    template="I want to open a restaurant for {cuisine} food. Suggest only one fancy name for this."
)

chain = prompt_template_name | llm
response = chain.invoke({"cuisine": "Indian"})
restaurant_name = response.content.strip()
print(response.content)


prompt_template_items = PromptTemplate(
    input_variables=["restaurant_name"],
    template="Suggest some menu items for {restaurant_name}. Return it as a comma separated"
)
chain = prompt_template_items | llm
response = chain.invoke({"restaurant_name": restaurant_name})

print("Menu Items:", response.content)
```
